# Test de rachas
Dada la muestra $u_1,...,u_n$ se construye una sucesión formada por $0$ y $1$ del siguiente modo: en la posición $i$ se coloca $0$ si $x_{i+1} < x_i$ y se coloca $1$ si $x_{i+a} > x_i$. Cada grupo consecutivo de $0$'s o $1$'s se denomina racha. Se demuestra que para $n$ suficientemente grande, el número de rachas $R$ sigue una distribución normal

$$ R \sim \mathcal{N}\left( \frac{2n-1}{3}, \sigma^2 = \frac{16n-29}{90} \right) $$

El estadístico de contraste sería

$$ Z = \frac{ R-\frac{2n-1}{3} }{ \sqrt{\frac{16n-29}{90} } } $$

que sigue una distribución $\mathcal{N}(0,1)$.
Para un nivel de significancia $\alpha$, se rechaza la hipótesis de aleatoriedad si $|Z|>Z_{\alpha/2}$.

# Generador congruencial mixto

```python
Xant = 43287492    # semilla

def genRand():
    global Xant
    a = 134775813
    c = 1
    m = 2**32
    Xnext = (a * Xant + c) % m
    Xant = Xnext
    U = Xnext / m
    return U

def genRandN(N):
    U=[]
    for t in range(1, N+1):
        ui = genRand()
        U.append(ui)
    return U
```

# Parámetros

```python
N = 100     # cantidad de números aleatorios a generar
```

Generar $n$ números aleatorios $u_1, u_2,...,u_n$, donde $n$ sea par,

```python
U = genRandN(N)
#U = np.random.uniform(0,1,N)

print("U =" , U[0:18])
```

```python
A = np.zeros(N-1)

for i in range(1,N):
    if U[i-1] > U[i]:
        A[i-1] = 0
    else:
        A[i-1] = 1

print(A[0:36])
```

```python
#import itertools

# se convierte el arreglo a booleano
A2 = np.array((A==1))
```

```python
# indica si hubo cambio de signo
sw = (A2[:-1] ^ A2[1:])
# indica las posiciones donde hubo cambio de signo
isw = np.arange(len(sw))[sw]
# se inserta la posición inicial -1
isw = list(isw)
isw.insert(0,-1)
isw = np.array(isw)
# se calcula la longitud de las rachas
L = isw[1:(len(isw))] - isw[0:(len(isw)-1)]
print(L[0:36])
```

```python
# se calcula entonces el estadístico de contraste Z a partir de el número de rachas R
R = len(L)
print("R = ",R)

Z = ( R - ((2*N-1)/3) )/ (np.sqrt( (16*N - 29)/(90) ))
print("Z = ",Z)
```

```python
# se calcula Z_{\alpha/2}
from scipy import stats
alpha = 0.05

Zt = stats.norm.ppf((1+(1-alpha))/2)

print("Z_{alpha/2} = ", Zt)

if abs(Z) > Zt:
    print("Se rechaza la hipotesis de aleatoriedad.")
else:
    print("Se acepta la hipotesis de aleatoriedad.")
```
